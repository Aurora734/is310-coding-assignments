```python
import pandas as pd

governors_df = pd.read_csv("https://raw.githubusercontent.com/melaniewalsh/responsible-datasets-in-context/main/datasets/gubernatorial-bios/gubernatorial_bios_final.csv", sep=',')

governors_df.info()
```

```python
governors_df.describe()
```

I'm intrigued by age_at_start and first_year.

```python
import altair as alt

governors_df['first_year'] = pd.to_datetime(governors_df['first_year'], format='%Y')

alt.Chart(governors_df).mark_point(size=60).encode(
    alt.X('first_year').scale(zero=False).title('First Year in Office'),
    # alt.X('first_year').scale(zero=False).axis(format='time%Y').title('First Year in Office'),
    alt.Y('age_at_start').title('Age at Start'),
    color='race/ethnicity',
    shape='gender',
    tooltip=['governor', 'state_territory', 'party', 'race/ethnicity', 'gender', 'first_year', 'age_at_start']
).properties(
    width=750,
    height=500
)
```

As expected, the majority of U.S. governors have been White men, and the governors who have been non-White or female skew heavily towards the past 70 or so years. Unexpectedly, there have actually been a good number of governors who entered office below the age of 40, including one governor below the age of 30 (Henry Clay Warmoth, who became governor at age 26 in 1868). I was expecting the ages to skew much further towards seniors, rather than being relatively distributed from 30s-60s, due to the recent notion that politicians are almost all senior citizens. However, this notion still makes a lot of sense when examining this graph seeing as there is a visible upwards trend in recent governors' ages at the start of their tenures. Since 2020, there has only been one new governor below the age of 45, whereas previously, there had been many new governors below the age of 45 across the history of the U.S.
